3.<< What are the different ways to define and call a function in Python? Ans<< Defining Functions in Python:

1. Simple Function: Define a function using the def keyword.
2. Function with Default Arguments: Define a function with default values for some or all parameters.
3. Function with Variable Arguments: Define a function that accepts a variable number of arguments.
4. Function with Keyword Arguments: Define a function that accepts keyword arguments.

Calling Functions in Python:
1. Positional Arguments: Call a function with positional arguments.
2. Keyword Arguments: Call a function with keyword arguments.
3. Mixing Positional and Keyword Arguments: Call a function with both positional and keyword arguments.
4. Unpacking Arguments: Call a function with unpacked arguments.
5. Unpacking Keyword Arguments: Call a function with unpacked keyword arguments.




















```python
#examples
#Simple Function:
def greet(name):
    print(f"Hello, {name}!")

#Function with Default Arguments:
def greet(name, msg="Hello"):
    print(f"{msg}, {name}!")

#Function with Variable Arguments:
def greet(*names):
    for name in names:
        print(f"Hello, {name}!")

#Function with Keyword Arguments:
def greet(**kwargs):
    for name, msg in kwargs.items():
        print(f"{msg}, {name}!")

#Calling Functions
#Positional Arguments:
greet("John")

#Keyword Arguments:

greet(name="John")

#Mixing Positional and Keyword Arguments:


def greet(name, msg):
    print(f"{msg}, {name}!")
greet("John", msg="Hello")

#Unpacking Arguments:
def greet(name, age):
    print(f"Hello, {name}! You are {age} years old.")
args = ["John", 30]
greet(*args)

#Unpacking Keyword Arguments:


def greet(name, age):
    print(f"Hello, {name}! You are {age} years old.")
kwargs = {"name": "John", "age": 30}
greet(**kwargs)


```

6.<<. Explain the concept of generators in Python and how they are defined?  Ans<<. Generators in Python
What are Generators?
- A type of iterable, like lists or tuples
- Can be used to generate a sequence of results instead of computing them all at once
- Useful for handling large datasets or infinite sequences

Defining Generators
- Generator Functions: Defined using the def keyword with a yield statement instead of return.
- Generator Expressions: Defined using a syntax similar to list comprehensions, but with parentheses instead of brackets.


```python
#example
# Generator Function
def infinite_sequence():
    num = 0
    while True:
        yield num
        num += 1

# Generator Expression
gen_expr = (x**2 for x in range(5))

# Usage
gen = infinite_sequence()
print(next(gen))
print(next(gen))

for value in gen_expr:
    print(value)

```

1. Memory Efficiency: Generators use less memory, as they generate values on-the-fly instead of storing them all at once.

10.<<  What is the difference between `map()`, `reduce()`, and `filter()` functions in Python? Ans.<< Differences between map(), reduce(), and filter()
map()
- Purpose: Applies a function to each item of an iterable and returns a map object.
- Output: A new iterable with transformed values.

filter()
- Purpose: Filters items from an iterable based on a condition and returns a filter object.
- Output: A new iterable with only the items that meet the condition.

reduce()
- Purpose: Applies a function to the first two items of an iterable, then to the result and the next item, and so on, reducing the iterable to a single value.
- Output: A single value, which is the result of the reduction operation.
- Note: reduce() is part of the functools module in Python 3.






```python
#example
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# map()
squared_numbers = list(map(lambda x: x**2, numbers))
print(squared_numbers)

# filter()
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)

# reduce()
sum_of_numbers = reduce(lambda x, y: x + y, numbers)
print(sum_of_numbers)

```

4.<< Write a Python function that checks if a given number is prime or not from 1 to 200

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def check_primes_in_range(start, end):
    primes = [num for num in range(start, end + 1) if is_prime(num)]
    return primes

start = 1
end = 200
primes = check_primes_in_range(start, end)
print("Prime numbers between", start, "and", end, "are:")
print(primes)

```

5.<<  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

```python
class FibonacciIterator:
    def __init__(self, n):
        self.n = n
        self.a = 0
        self.b = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.n:
            fib_num = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return fib_num
        else:
            raise StopIteration

n = 10
fib_iterator = FibonacciIterator(n)
for num in fib_iterator:
    print(num)

```

6.<< Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i

n = 5
for power in powers_of_two(n):
    print(power)

```

8.<< Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

```python
def get_tuples_from_user():
    n = int(input("Enter the number of tuples: "))
    tuples_list = []
    for i in range(n):
        tup = tuple(map(int, input(f"Enter tuple {i+1} (space-separated): ").split()))
        tuples_list.append(tup)
    return tuples_list

def sort_tuples(tuples_list):
    return sorted(tuples_list, key=lambda x: x[1])

def main():
    tuples_list = get_tuples_from_user()
    print("Original list of tuples:")
    print(tuples_list)

    sorted_tuples = sort_tuples(tuples_list)
    print("Sorted list of tuples based on the second element:")
    print(sorted_tuples)

if __name__ == "__main__":
    main()
```

9.<< Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

def get_temps_from_user():
    temps = input("Enter temperatures in Celsius (space-separated): ")
    try:
        celsius_temps = list(map(float, temps.split()))
        return celsius_temps
    except ValueError:
        print("Invalid input. Please enter numbers separated by spaces.")
        return get_temps_from_user()

def main():
    celsius_temps = get_temps_from_user()
    print("Celsius temperatures:", celsius_temps)

    fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))
    print("Fahrenheit temperatures:", fahrenheit_temps)

if __name__ == "__main__":
    main()


```

10.<<  Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
def is_not_vowel(char):
    return char.lower() not in 'aeiou'

def main():
    input_str = input("Enter a string: ")
    chars = list(input_str)
    non_vowel_chars = list(filter(is_not_vowel, chars))
    result_str = ''.join(non_vowel_chars)
    print("String without vowels:", result_str)

if __name__ == "__main__":
    main()
```
